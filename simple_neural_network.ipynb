```python
import numpy as np

class SimpleNeuralNetwork:
    def __init__(self, input_size):
        self.weights = np.random.rand(input_size, 1)

    @staticmethod
    def sigmoid(x):
        return 1/(1+np.exp(-x))

    @staticmethod
    def sigmoid_derivative(x):
        return x*(1-x)

    def predict(self, X):
        return self.sigmoid(np.matmul(X, self.weights))

    def train(self, X, y, epochs=1000):
        X, Y = np.asarray(X), np.asarray([y]).T
        for _ in range(epochs):
            Ycap = self.predict(X)
            error = Y-Ycap
            self.weights += np.matmul(X.T, error*self.sigmoid_derivative(Ycap))
```

```python
X = [[1, 1],
     [1, 0],
     [0, 1],
     [0, 0]]
y = [1, 1, 1, 0]

snn = SimpleNeuralNetwork(2)
snn.train(X, y, epochs=2000)
print(snn.weights)
print(snn.predict([0, 1]))
```
