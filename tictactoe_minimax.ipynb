```python
def draw(board):
    print('{:^5}|{:^5}|{:^5}\n-----------------\n{:^5}|{:^5}|{:^5}\n-----------------\n{:^5}|{:^5}|{:^5}\n'.format(*board))

def evaluate(board):
	for i in range(0, 9, 3):
		if board[i]==board[i+1]==board[i+2]!='':
			return 10 if board[i]=='x' else -10
	for i in range(3):
		if board[i]==board[i+3]==board[i+6]!='':
			return 10 if board[i]=='x' else -10
	if board[0]==board[4]==board[8]!='':
		return 10 if board[0]=='x' else -10
	if board[2]==board[4]==board[6]!='':
		return 10 if board[2]=='x' else -10
	
	if all(board):
		return 0
	
	return None

def minimax(board, is_max=True, val=False):
	score = evaluate(board)
	if score is not None:
		return score
	
	best_move = -1
	best_val, func, mark = (-1000, max, 'x') if is_max else (1000, min, 'o')
	for i in range(9):
		if board[i]=='':
			board[i] = mark
			move_val = minimax(board, not is_max, val=True)
			board[i] = ''
			if best_val!=func(best_val, move_val):
				best_val = move_val
				best_move = i
	
	return best_val if val else best_move

def play(user, is_max, board):
	if user:
		move = int(input('Enter the position to mark: '))-1
		if board[move]!='':
			raise ValueError('marked positions are not allowed to mark')
	else:
		move = minimax(board, is_max)
	
	board[move] = 'x' if is_max else 'o'

def start(user=True):
	board = ['']*9
	is_max = True
	winner = None
	while winner is None:
		play(user, is_max, board)
		draw(board)
		winner = evaluate(board)
		is_max = not is_max
		user = not user
	
	if winner==10: print('x won')
	elif winner==-10: print('o won')
	else: print('tie')
```

```python
start()
```
