```python
from simpleai.search import CspProblem, backtrack, min_conflicts

variables = ('WA', 'NT', 'SA', 'Q', 'NSW', 'V', 'T')
domains = {var: ['red', 'green', 'blue'] for var in variables}

def const_different(variables, values):
    return values[0]!=values[1]

constraints = [
    (('WA', 'NT'), const_different),
    (('WA', 'SA'), const_different),
    (('SA', 'NT'), const_different),
    (('SA', 'Q'), const_different),
    (('NT', 'Q'), const_different),
    (('SA', 'NSW'), const_different),
    (('Q', 'NSW'), const_different),
    (('SA', 'V'), const_different),
    (('NSW', 'V'), const_different)
    ]

my_problem = CspProblem(variables, domains, constraints)
print(backtrack(my_problem))
print(backtrack(my_problem, value_heuristic='lvc'))
print(backtrack(my_problem, variable_heuristic='mcv'))
print(backtrack(my_problem, variable_heuristic='degree'))
print(backtrack(my_problem, variable_heuristic='mcv', value_heuristic='lvc'))
print(backtrack(my_problem, variable_heuristic='degree', value_heuristic='lvc'))
print(min_conflicts(my_problem))
```
