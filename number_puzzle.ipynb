```python
from simpleai.search import SearchProblem, astar
```

```python
GOAL = '''1-2-3
4-5-6
7-8-e'''

INITIAL = '''4-1-2
7-e-3
8-5-6'''
```

```python
def mat_to_string(mat):
    return '\n'.join(['-'.join(row) for row in mat])

def string_to_mat(string):
    return [row.split('-') for row in string.split('\n')]
```

```python
def find_location(mat, to_find):
    for i, row in enumerate(mat):
        for j, number in enumerate(row):
            if number==to_find:
                return i, j
```

```python
goal_positions = {}
goal_mat = string_to_mat(GOAL) 
for number in '12345678e':
    goal_positions[number] = find_location(goal_mat, number)
```

```python
class EightPuzzleProblem(SearchProblem):
    def actions(self, state):
        mat = string_to_mat(state)
        row_e, col_e = find_location(mat, 'e')
        actions = []
        if row_e>0:
            actions.append(mat[row_e-1][col_e])
        if row_e<2:
            actions.append(mat[row_e+1][col_e])
        if col_e>0:
            actions.append(mat[row_e][col_e-1])
        if col_e<2:
            actions.append(mat[row_e][col_e+1])
        
        return actions
    
    def result(self, state, action):
        mat = string_to_mat(state)
        row_e, col_e = find_location(mat, 'e')
        row_n, col_n = find_location(mat, action)
        mat[row_e][col_e], mat[row_n][col_n] = mat[row_n][col_n], mat[row_e][col_e]

        return mat_to_string(mat)
    
    def is_goal(self, state):
        return state==GOAL
    
    def cost(self, state1, action, state2):
        return 1
    
    def heuristic(self, state):
        mat = string_to_mat(state)
        distance = 0
        for number in '12345678e':
            row_n, col_n = find_location(mat, number)
            row_n_goal, col_n_goal = goal_positions[number]
            distance += abs(row_n-row_n_goal)+abs(col_n-col_n_goal)
        
        return distance
```

```python
result = astar(EightPuzzleProblem(INITIAL))
for action, state in result.path():
    print('Move number:', action)
    print(state)
```
