```python
graph = {
    'A': {'B', 'C'},
    'B': {'A', 'D', 'E'},
    'C': {'A', 'F'},
    'D': {'B'},
    'E': {'B', 'F'},
    'F': {'C', 'E'},
}
```

```python
def _dfs(graph, node, visited):
    if node not in visited:
        visited.append(node)
        for neighbour in graph[node]:
            _dfs(graph, neighbour, visited)

def dfs(graph, node):
    _dfs(graph, node, visited:=[])
    
    return visited
```

```python
dfs(graph, 'A')
```
