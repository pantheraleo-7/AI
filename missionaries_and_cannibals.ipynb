```python
def display(p1, p2, rm, rc, um, uc, rside):
    print('M'*rm, end=' ')
    print('C'*rc, end=' ')
    if rside:
        print(f'|------water------\\{p1} , {p2}/--|', end=' ')
    else:
        print(f'|--\\{p1} , {p2}/------water------|', end=' ')
    print('M'*um, end=' ')
    print('C'*uc)

def isreached(rm, rc):
    return rm==3==rc

def solve():
    um = uc = 3
    rm = rc = 0
    rside = True
    to_travel = 0
    display(' ', ' ', rm, rc, um, uc, rside)
    while not isreached(rm, rc):
        if rside:
            if to_travel==1:
                display('C', ' ', rm, rc, um, uc, rside)
                uc += 1
            elif to_travel==2:
                display('C', 'M', rm, rc, um, uc, rside)
                um += 1
                uc += 1
            if um-2>=uc and rm+2>=rc or um-2==0:
                um -= 2
                display('M', 'M', rm, rc, um, uc, rside)
                to_travel = 1
            elif uc-2<um and rm==0 or rc+2<=rm or um==0:
                uc -= 2
                display('C', 'C', rm, rc, um, uc, rside)
                to_travel = 2
            elif uc-1<=um-1 and rm+1>=rc+1:
                um -= 1
                uc -= 1
                display('M', 'C', rm, rc, um, uc, rside)
                to_travel = 3
            rside = False
        else:
            if to_travel==1:
                display('M', 'M', rm, rc, um, uc, rside)
                rm += 2
            elif to_travel==2: 
                display('C', 'C', rm, rc, um, uc, rside)
                rc += 2 
            elif to_travel==3:
                display('M', 'C', rm, rc, um, uc, rside)
                rc += 1
                rm += 1
            if not isreached(rm, rc):
                if rc>1 and rm==0 or um==0:
                    rc -= 1
                    display('C', ' ', rm, rc, um, uc, rside)
                    to_travel = 1
                elif uc+2>um:
                    rm -= 1
                    rc -= 1
                    display('C', 'M', rm, rc, um, uc, rside)
                    to_travel = 2
                rside = True
    display(' ', ' ', rm, rc, um, uc, rside)
```

```python
solve()
```

```python
def move_tower(height, from_pole, to_pole, with_pole):
    if height>=1:
        move_tower(height-1, from_pole, with_pole, to_pole)
        print(f'moving disk from {from_pole} to {to_pole}')
        move_tower(height-1, with_pole, to_pole, from_pole)
```
