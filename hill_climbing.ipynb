```python
def dist(x1, y1, x2, y2):
    return (x2-x1)**2 + (y2-y1)**2

def sum_dists(x, y, point1, point2, point3, point4):
    d1 = dist(x, y, *point1)
    d2 = dist(x, y, *point2)
    d3 = dist(x, y, *point3)
    d4 = dist(x, y, *point4)
    return d1+d2+d3+d4

def point_with_dist(x, y, point1, point2, point3, point4):
    return [x, y, sum_dists(x, y, point1, point2, point3, point4)]

def hill_climb(start_point, point1, point2, point3, point4, step):
    i = 0
    x, y = start_point
    min_dist = sum_dists(x, y, point1, point2, point3, point4)
    while True:
        i += 1
        print(f'{i:3}: {x:^8.2f} {y:^8.2f}')
        p1 = point_with_dist(x+step, y, point1, point2, point3, point4)
        p2 = point_with_dist(x-step, y, point1, point2, point3, point4)
        p3 = point_with_dist(x, y+step, point1, point2, point3, point4)
        p4 = point_with_dist(x, y-step, point1, point2, point3, point4)
        *new_point, new_min_dist = min(p1, p2, p3, p4, key=lambda p: p[2])
        if new_min_dist<min_dist:
            x, y = new_point
            min_dist = new_min_dist
        else: break
```

```python
hill_climb([1,1], [1,5], [6, 4], [5, 2], [2, 1], 0.1)
```
