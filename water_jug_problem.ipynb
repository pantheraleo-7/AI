```python
capacity = (12, 8, 5)
x, y, z = capacity
memory = {}
soln = []
```

```python
def get_all_states(state):
    a, b, c = state

    if a==6==b:
        soln.append(state)
        return True
    
    if state in memory:
        return False
    
    memory[state] = 1
    
    if a>0:
        if a+b<=y:
            if get_all_states((0, a+b, c)):
                soln.append(state)
                return True
        elif get_all_states((a-(y-b), y, c)):
            soln.append(state)
            return True
            
        if a+c<=z:
            if get_all_states((0, b, a+c)):
                soln.append(state)
                return True
        elif get_all_states((a-(z-c), b, z)):
            soln.append(state)
            return True
    
    if b>0:
        if a+b<=x:
            if get_all_states((a+b, 0, c)):
                soln.append(state)
                return True
        elif get_all_states((x, b-(x-a), c)):
            soln.append(state)
            return True
            
        if b+c<=z:
            if get_all_states((a, 0, b+c)):
                soln.append(state)
                return True
        elif get_all_states((a, b-(z-c), z)):
            soln.append(state)
            return True

    if c>0:
        if a+c<=x:
            if get_all_states((a+c, b, 0)):
                soln.append(state)
                return True
        elif get_all_states((x, b, c-(x-a))):
            soln.append(state)
            return True
            
        if b+c<=y:
            if get_all_states((a, b+c, 0)):
                soln.append(state)
                return True
        elif get_all_states((a, y, c-(y-b))):
            soln.append(state)
            return True
    
    return False
```

```python
initial_state = (12, 0, 0)
print('Starting work')
get_all_states(initial_state)
soln.reverse()
for i in soln:
    print(i)
```
