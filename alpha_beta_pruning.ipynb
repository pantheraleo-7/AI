```python
def alpha_beta(tree, maximiser=True, alpha=float('-inf'), beta=float('+inf')):
    if isinstance(tree, int):
        return tree
    
    for branch in tree:
        val = alpha_beta(branch, not maximiser, alpha, beta)
        if maximiser:
            alpha = max(alpha, val)
        else:
            beta = min(beta, val)
        
        if alpha>=beta:
            break
    
    return alpha if maximiser else beta
```

```python
tree = [[[5, 1, 2],
         [8, -8, -9]],
         [[9, 4, 5],
          [-3, 4, 3]]]
```

```python
alpha_beta(tree)
```
